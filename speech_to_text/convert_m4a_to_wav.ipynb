```python
import os
import subprocess

# Directory containing your .m4a files
input_dir = "/work/speech/interviews/"  # change this if needed
output_dir = input_dir + "wav/"  # output directory for converted files

os.makedirs(output_dir, exist_ok=True)

# Loop over all .m4a files in the directory
for filename in os.listdir(input_dir):
    if filename.lower().endswith(".m4a"):
        input_path = os.path.join(input_dir, filename)
        output_path = os.path.join(output_dir, os.path.splitext(filename)[0] + ".wav")

        # ffmpeg command: convert to mono, 16kHz WAV (common for speech models)
        command = [
            "ffmpeg",
            "-y",  # overwrite existing
            "-i", input_path,
            "-ac", "1",          # mono
            "-ar", "16000",      # 16 kHz
            output_path
        ]

        print(f"Converting: {filename} → {output_path}")
        subprocess.run(command, stdout=subprocess.DEVNULL, stderr=subprocess.DEVNULL)

print("✅ Conversion complete!")
```
